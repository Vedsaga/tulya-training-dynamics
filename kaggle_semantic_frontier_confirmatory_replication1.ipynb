# Tulya Semantic Frontier — Confirmatory Replication 1

This notebook runs the **fresh-seed confirmatory replication** of the preregistered generality-premium test.

It does not modify the original falsification experiment. The design and decision rule are frozen in `SEMANTIC_FRONTIER_CONFIRMATORY_REPLICATION1_PREREGISTRATION.md`.

Only seeds **10–39** are used. Seeds 0–9 from the original run are excluded from the replication decision.

```python
from pathlib import Path
import importlib.util, requests, sys

REPO = "https://raw.githubusercontent.com/Vedsaga/tulya-training-dynamics/main"
WORK = Path("/kaggle/working")

files = [
    "semantic_frontier_v2.py",
    "semantic_frontier_confirmatory_replication1.py",
]

for name in files:
    path = WORK / name
    url = f"{REPO}/{name}"
    r = requests.get(url, timeout=30)
    r.raise_for_status()
    path.write_bytes(r.content)
    print("Downloaded", url)

# Load the base module first so the replication module imports the exact file we downloaded.
spec = importlib.util.spec_from_file_location(
    "semantic_frontier_v2", WORK / "semantic_frontier_v2.py"
)
sf = importlib.util.module_from_spec(spec)
sys.modules["semantic_frontier_v2"] = sf
spec.loader.exec_module(sf)

spec2 = importlib.util.spec_from_file_location(
    "semantic_frontier_confirmatory_replication1",
    WORK / "semantic_frontier_confirmatory_replication1.py",
)
rep = importlib.util.module_from_spec(spec2)
sys.modules["semantic_frontier_confirmatory_replication1"] = rep
spec2.loader.exec_module(rep)

print("Device:", sf.DEVICE)
print("Replication seeds:", rep.SEEDS[0], "through", rep.SEEDS[-1])
print("Capacities:", rep.CAPACITIES)
```

```python
runs, pairs, analysis = rep.run_replication()
analysis
```

```python
print("Decision:", analysis["decision"])
print("Manipulation passed:", analysis["manipulation_check_passed"])
print("Qualified 4-vs-6 pairs:", analysis["qualified_pair_count_4_vs_6"])
print("Median delta excess:", analysis["median_delta_excess_bits_6minus4"])
print("Median delta future uncertainty:", analysis["median_delta_future_uncertainty_6minus4"])
print("Paired improvements:", analysis["paired_future_improvement_count"],
      "/", analysis["qualified_pair_count_4_vs_6"])
print("Spearman rho:", analysis["spearman_excess_vs_future_uncertainty"])

display(pairs)
```
