# Tulya Semantic Frontier — preregistered falsification run

This notebook runs the frozen primary test in `SEMANTIC_FRONTIER_FALSIFICATION_PREREGISTRATION.md`. It does **not** change thresholds after seeing results.

Primary question: among representations that are already approximately sufficient for the current tasks, does retaining more excess information `H(Z|S)` reduce expected uncertainty about unseen future tasks?

The experiment holds tasks, data size, model family, optimizer, and compute budget fixed; it varies only binary bottleneck capacity (4/5/6 bits) and random seed (0–9).

```python
from pathlib import Path
import importlib.util, requests, sys, json

def load_raw(module_name, filename):
    path=Path('/kaggle/working')/filename
    url=f'https://raw.githubusercontent.com/Vedsaga/tulya-training-dynamics/main/{filename}'
    r=requests.get(url,timeout=30); r.raise_for_status(); path.write_bytes(r.content)
    spec=importlib.util.spec_from_file_location(module_name,path)
    mod=importlib.util.module_from_spec(spec); sys.modules[module_name]=mod; spec.loader.exec_module(mod)
    print('Loaded',filename)
    return mod

sf=load_raw('semantic_frontier_v2','semantic_frontier_v2.py')
ff=load_raw('semantic_frontier_falsification','semantic_frontier_falsification.py')
print('device:',sf.DEVICE)
```

## Run the frozen primary test

Expected workload: 30 runs = 3 bottleneck capacities × 10 seeds, each with the same 1024 training states and 3000-step budget. The script automatically applies the preregistered qualification, manipulation check, and support/falsification decision rule.

```python
runs,pairs,analysis=ff.run_primary()
analysis
```
